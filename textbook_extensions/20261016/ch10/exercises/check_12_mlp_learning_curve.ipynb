입력 특성 행렬(Feature Matrix, X)를 오른쪽 식으로 계산해 다음 단계에서 사용합니다. 값과 형태(Shape)를 구분하고 앞 셀의 입력이 준비되어 있는지 확인합니다.

목표값(Target, y)를 오른쪽 식으로 계산해 다음 단계에서 사용합니다. 값과 형태(Shape)를 구분하고 앞 셀의 입력이 준비되어 있는지 확인합니다.

훈련 평가 분리(Train Test Split)를 수행합니다. 행 번호 두 배열이면 앞은 훈련, 뒤는 평가이며 네 배열이면 입력과 목표의 행 대응을 유지합니다. test_size는 평가 비율이고 stratify는 클래스 비율 유지 조건입니다.

변수 rng를 오른쪽 식으로 계산해 다음 단계에서 사용합니다. 값과 형태(Shape)를 구분하고 앞 셀의 입력이 준비되어 있는지 확인합니다.

변수 ordered를 오른쪽 식으로 계산해 다음 단계에서 사용합니다. 값과 형태(Shape)를 구분하고 앞 셀의 입력이 준비되어 있는지 확인합니다.

Stratified prefixes ensure every class is represented at small sizes.

반복문(Loop)의 조건에 따라 들여쓴 계산을 수행합니다. for group in zip(*[list(a)+[None]*(max(map(len,by_class))-len(a)) for a in by_class]):에서 반복 대상과 횟수를 확인하고 종료 뒤에는 마지막 계산 상태가 남는다는 점에 주의합니다.

변수 rows를 오른쪽 식으로 계산해 다음 단계에서 사용합니다. 값과 형태(Shape)를 구분하고 앞 셀의 입력이 준비되어 있는지 확인합니다.

반복문(Loop)의 조건에 따라 들여쓴 계산을 수행합니다. for size in [18,36,60,90,len(pool)]:에서 반복 대상과 횟수를 확인하고 종료 뒤에는 마지막 계산 상태가 남는다는 점에 주의합니다.

```python
# 쉼표 구분 자료(Comma Separated Values, CSV)를 읽습니다. 이 실습에서 선정한 입력 파일과 열 이름·목표 열을 확인합니다. 입력 자료의 선정 이유와 실제 행 수는 앞의 자료 설명을 따릅니다.
f=pd.read_csv(DATA/"wine.csv")

# 입력 특성 행렬(Feature Matrix, X)를 오른쪽 식으로 계산해 다음 단계에서 사용합니다. 값과 형태(Shape)를 구분하고 앞 셀의 입력이 준비되어 있는지 확인합니다.
X=f.drop(columns="target")

# 목표값(Target, y)를 오른쪽 식으로 계산해 다음 단계에서 사용합니다. 값과 형태(Shape)를 구분하고 앞 셀의 입력이 준비되어 있는지 확인합니다.
y=f.target

# 훈련 평가 분리(Train Test Split)를 수행합니다. 행 번호 두 배열이면 앞은 훈련, 뒤는 평가이며 네 배열이면 입력과 목표의 행 대응을 유지합니다. test_size는 평가 비율이고 stratify는 클래스 비율 유지 조건입니다.
pool,va=train_test_split(np.arange(len(f)),test_size=.3,stratify=y,random_state=42)

# 변수 rng를 오른쪽 식으로 계산해 다음 단계에서 사용합니다. 값과 형태(Shape)를 구분하고 앞 셀의 입력이 준비되어 있는지 확인합니다.
rng=np.random.default_rng(42)

# 변수 ordered를 오른쪽 식으로 계산해 다음 단계에서 사용합니다. 값과 형태(Shape)를 구분하고 앞 셀의 입력이 준비되어 있는지 확인합니다.
ordered=[]

# Stratified prefixes ensure every class is represented at small sizes.
by_class=[rng.permutation(pool[y.iloc[pool].to_numpy()==c]) for c in sorted(y.unique())]

# 반복문(Loop)의 조건에 따라 들여쓴 계산을 수행합니다. for group in zip(*[list(a)+[None]*(max(map(len,by_class))-len(a)) for a in by_class]):에서 반복 대상과 횟수를 확인하고 종료 뒤에는 마지막 계산 상태가 남는다는 점에 주의합니다.
for group in zip(*[list(a)+[None]*(max(map(len,by_class))-len(a)) for a in by_class]):
    ordered.extend([i for i in group if i is not None])

# 변수 rows를 오른쪽 식으로 계산해 다음 단계에서 사용합니다. 값과 형태(Shape)를 구분하고 앞 셀의 입력이 준비되어 있는지 확인합니다.
rows=[]

# 반복문(Loop)의 조건에 따라 들여쓴 계산을 수행합니다. for size in [18,36,60,90,len(pool)]:에서 반복 대상과 횟수를 확인하고 종료 뒤에는 마지막 계산 상태가 남는다는 점에 주의합니다.
for size in [18,36,60,90,len(pool)]:
    tr=np.array(ordered[:size])
    m=make_pipeline(StandardScaler(),MLPClassifier(hidden_layer_sizes=(12,),
        solver="lbfgs",alpha=.2,max_iter=2000,random_state=42))
    m.fit(X.iloc[tr],y.iloc[tr])
    rows.append({"train_n":len(tr),"train_accuracy":accuracy_score(y.iloc[tr],m.predict(X.iloc[tr])),
        "validation_accuracy":accuracy_score(y.iloc[va],m.predict(X.iloc[va]))})
```
